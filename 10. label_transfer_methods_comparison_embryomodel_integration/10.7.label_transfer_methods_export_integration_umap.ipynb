```python
# Define the color palette as a dictionary
lineage_color = {
    "Amniotic_ecto": "#1f77b4",    # Blue
    "Notochord": "#aa40fc",        # Purple
    "Endoderm": "#ff7f0e",         # Orange
    "PGC": "#8c564b",             # Brown
    "ExE_endo": "#279e68",        # Green
    "Primitive.streak": "#e377c2", # Pink
    "NMP": "#d62728",             # Red
    "TE_TrB": "#b5bd61",          # Yellow-Green
    "epi": "#17becf",             # Cyan
    "hemogenic": "#aec7e8",       # Light Blue
    "meso_Exe.meso": "#ffbb78",   # Orange-Yellow
    "neural_ecto": "#98df8a"      # Green-Yellow
}
```

```python
# Define list of genes
genes_to_plot = ["POU5F1","NANOG",#epiblast
                 "SOX2", "TTYH1", #Neural_ectoderm
                 "GATA3","TFAP2A",
                 "TBXT",  "CDX1","PDGFRA","APOA2","FOXA2","NANOS3",
                "PECAM1","HBZ","PTPRC",
                "GABRP","HEY1","COL6A1","COL6A2",]  # Replace "GENE2" and "GENE3" with the actual gene names
```

```python
adata = sc.read_h5ad("embryo_model_integration_scPoli.h5ad") 
# Set colors in adata.uns
adata.uns['lineage_color'] = lineage_color

adata
```

```python

sc.pl.umap(adata, color=['human_ref_lineage_pred'], palette=lineage_color, save="scpoli_lineage.pdf")
sc.pl.umap(adata, color=['orig.ident'],save="scpoli_orig_ident.pdf")

# Directory to save plots
output_dir = "./gene_expression_plots_scpoli"
import os
os.makedirs(output_dir, exist_ok=True)
# Loop over each gene and plot/save individually
for gene in genes_to_plot:
    fig, ax = plt.subplots(figsize=(6, 5))  # Create a new figure for each gene
    sc.pl.umap(
        adata,
        color=gene,
        use_raw=False,
        cmap=sns.cubehelix_palette(dark=0, light=.9, as_cmap=True),
        ax=ax,
        show=False,  # Prevents Scanpy from calling plt.show()
        size=15
    )
    plt.tight_layout()
    plt.savefig(f"{output_dir}/{gene}_umap.pdf", dpi=300, bbox_inches='tight')
    plt.close()
```
